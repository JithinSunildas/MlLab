```python
df=pd.read_csv("house_price.csv")

df.head()
df.info()
df.describe()
print(df.head())
print(df.info())
print(df.describe())
```

```python
df.fillna(df.select_dtypes(include=['number']).mean(), inplace=True)
df.dropna(inplace = True)
```

```python
Q1= df["price"].quantile(.25)
Q3= df["price"].quantile(.75)
iqr= Q3- Q1

l_limit= Q1- 1.5* iqr
u_limit= Q3+ 1.5* iqr

df= df[(df["price"]>= l_limit) & (df["price"]<= u_limit) ]
df.describe()
```
